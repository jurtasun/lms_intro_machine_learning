**Supervised learning: classification**

We have cells of known type (A, B, C), each described by the expression of three genes, and want to classify a new cell.
We look at the $k$ cells most similar to it and give it the most common type among them (majority vote).

**The maths**

Similar cells are cells that are close to each other. The Euclidean distance between the new cell $x$ and a known cell $y$ adds up the squared differences in each gene, then takes the square root:

$$d(x, y) = \sqrt{(x_1 - y_1)^2 + (x_2 - y_2)^2 + (x_3 - y_3)^2}$$

KNN does not train any parameters: it only picks the $k$ cells with the **smallest** distance and takes a vote.

```python
# Import libraries 
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Generate example input 
print("\nGenerate example input")

# Example data: rows represent cells, columns represent features (e.g. gene expression)
data = np.array([
    [5.1, 1.4, 0.2], # Cell type A
    [4.9, 1.4, 0.2],
    [6.2, 5.4, 2.3], # Cell type B
    [5.9, 5.1, 1.8],
    [8.2, 5.4, 5.3], # Cell type C
    [7.9, 4.1, 4.8]])
print("\nData:\n", data)
print("Shape:", data.shape)
print("Format: ", type(data))
```

```python
# Labels for training: cell types
labels = np.array(['A', 'A', 'B', 'B', 'C', 'C'])
print("\nLabels: ", labels)
print("Shape:", labels.shape)
print("Format: ", type(labels))

# New cell to classify
# x_new = np.array([4.8, 1.0, 0.5])
# x_new = np.array([5.8, 5.0, 1.7])
x_new = np.array([7.8, 5.0, 5.5])
print("\nPoint to classify: ", x_new)
print("Shape:", x_new.shape)
print("Format: ", type(x_new))
```

```python
# Euclidean distance, step by step
# d = sqrt((a1 - b1)^2 + (a2 - b2)^2 + (a3 - b3)^2)
cell_0 = data[0]
print("Cell 0:   ", cell_0)
print("New cell: ", x_new)

# Difference in each coordinate
differences = cell_0 - x_new
print("Differences:", differences)

# Square each difference
squared = differences**2
print("Squared:", squared)

# Sum and take the square root
distance = np.sqrt(np.sum(squared))
print(f"Distance: {distance:.3f}")

# Check with numpy implementation
print(f"numpy:    {np.linalg.norm(cell_0 - x_new):.3f}")
```

```python
# KNN implementation 
print("\nKNN running")

def knn_predict(data, labels, x_new, k = 3):

    # Compute Euclidean distances from x_new to all cells
    # Same steps as above, now for all rows at once
    # Argument axis = 1 tells numpy to sum row by row
    differences = data - x_new
    distances = np.sqrt(np.sum(differences**2, axis = 1))
    print("Distance to cells in data:", distances)
    
    # Sort distances from smallest to largest
    sorted_indices = distances.argsort()
    # Take the first k indices
    knn_indices = sorted_indices[:k]
    print(f"Index of the {k} nearest neighbors:", knn_indices)
    
    # Get the labels of the nearest neighbors
    knn_labels = labels[knn_indices]
    print("Labels of nearest neighbors:", knn_labels)
    
    # Return the most common label (majority vote)
    # First [0]: Takes the first tuple in the list of most common elements (e.g. ('B', 2)).
    # Second [0]: Takes the label from the tuple (ignores the count).
    most_common = Counter(knn_labels).most_common(1)[0][0]

    return most_common

# Classify the new cell
label_new = knn_predict(data, labels, x_new, k = 3)
print(f"\nNew cell classified as: {label_new}")
```
