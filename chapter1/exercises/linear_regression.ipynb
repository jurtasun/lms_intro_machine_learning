**Supervised learning: regression**

We measure gene expression $y$ at different doses of a stimulus $x$, and want to predict the expression for a new dose.
We fit a straight line to the observations by hand, check it with numpy, and measure how far a line is from the data.

**The maths**

The model is a straight line, with slope $a$ and intercept $b$. For each dose $x$ it gives a prediction $\hat{y}$:

$$\hat{y} = a\,x + b$$

To decide how good a line is, we take the difference between each observation $y_i$ and its prediction $\hat{y}_i$, square it, and average over the $N$ observations. This is the mean squared error, and the best line is the one that makes it as **small** as possible:

$$\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} \left( y_i - \hat{y}_i \right)^2$$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Grid of x values: e.g. stimulus (drug dose)
x = np.linspace(0, 10, 11)
print("x:", x)
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# True function: y(x) = slope * x + intercept
# In real life we never know it, here we use it to simulate data
def f(x):
    return 0.5 * x + 1.2

# Observations: true function plus some random noise
np.random.seed(0)
y = f(x) + np.random.normal(0, 0.4, size = x.shape)
print("y:", y)
print("Shape:", y.shape)
```

```python
# Plot the true function and the observations
plt.figure(figsize = (8, 5))
plt.plot(x, f(x), color = 'gray', linestyle = 'dashed', label = 'True function')
plt.scatter(x, y, color = 'blue', label = 'Observations')
plt.xlabel('x: stimulus')
plt.ylabel('y(x): gene expression')
plt.legend()
plt.show()
```

```python
# Linear regression fit
# Slope and intercept that minimize the Residual Sum of Squares (RSS):
# slope = sum((x - x_mean) * (y - y_mean)) / sum((x - x_mean)^2)
# intercept = y_mean - slope * x_mean
x_mean = np.mean(x)
y_mean = np.mean(y)

slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean)**2)
intercept = y_mean - slope * x_mean
print(f"Fit: y = {slope:.2f} * x + {intercept:.2f}")
```

```python
# Check with numpy implementation
slope_np, intercept_np = np.polyfit(x, y, deg = 1)
print(f"numpy: y = {slope_np:.2f} * x + {intercept_np:.2f}")
```

```python
# Residuals: observed minus fitted values
y_fit = slope * x + intercept
residuals = y - y_fit
print("Residuals:", residuals)

rss = np.sum(residuals**2) # residual sum of squares
tss = np.sum((y - y_mean)**2) # total sum of squares
r2 = 1 - rss / tss
print(f"RSS: {rss:.2f}, R2: {r2:.2f}")
```

```python
# Predict y for new x values
x_new = np.array([1.5, 4.5, 7.5])
y_new = slope * x_new + intercept
print("Predictions for new x:", y_new)
```

```python
# Plot observations, fitted line, residuals and predictions
plt.figure(figsize = (8, 5))
plt.scatter(x, y, color = 'blue', label = 'Observations')
plt.plot(x, y_fit, color = 'red', label = 'Fitted line')
plt.scatter(x_new, y_new, color = 'green', marker = 'x', s = 80, label = 'Predictions')

# Residuals as vertical lines
for xi, yi, yfi in zip(x, y, y_fit):
    plt.vlines(xi, yfi, yi, color = 'gray', linestyle = 'dashed', alpha = 0.5)

plt.xlabel('x: stimulus')
plt.ylabel('y(x): gene expression')
plt.title('Linear regression')
plt.legend()
plt.show()
```

```python
# Mean squared error (MSE)
# How far is a given function from the observations?
# MSE = mean((y - y_pred)^2)
def mse(y, y_pred):
    return np.mean((y - y_pred)**2)

# A guess: a line with slope 1 and intercept 0
y_guess = 1.0 * x + 0.0
print(f"MSE of the guess: {mse(y, y_guess):.2f}")
print(f"MSE of the fit:   {mse(y, y_fit):.2f}")
```

```python
# Plot the guess on top of the observations
plt.figure(figsize = (8, 5))
plt.scatter(x, y, color = 'blue', label = 'Observations')
plt.plot(x, y_guess, color = 'orange', label = 'Guess: y = x')

for xi, yi, ygi in zip(x, y, y_guess):
    plt.vlines(xi, ygi, yi, color = 'gray', linestyle = 'dashed', alpha = 0.5)

plt.xlabel('x: stimulus')
plt.ylabel('y(x): gene expression')
plt.title(f'MSE = {mse(y, y_guess):.2f}')
plt.legend()
plt.show()
```

```python
# Try several slopes, keeping the fitted intercept
slopes = np.linspace(-0.5, 1.5, 21)
errors = [mse(y, s * x + intercept) for s in slopes]

for s, e in zip(slopes, errors):
    print(f"slope = {s:.1f}  ->  MSE = {e:.2f}")
```

```python
# MSE as a function of the slope
# The minimum is at the fitted slope: this curve is our first loss function
plt.figure(figsize = (8, 5))
plt.plot(slopes, errors, color = 'purple', marker = 'o')
plt.axvline(slope, color = 'red', linestyle = 'dashed', label = f'Fitted slope = {slope:.2f}')
plt.xlabel('slope')
plt.ylabel('MSE')
plt.title('MSE vs slope')
plt.legend()
plt.show()
```
