**Supervised learning: regression**

We measure gene expression $y$ at different doses of a stimulus $x$, and want the straight line that best describes the measurements.
The idea is always the same: the **input** $x$ goes into a **model** (the line), which gives an **output**, the prediction $y^{\text{pred}}$. We compare the output with the **label** $y$ (the measurement) using a **loss**, and the best line is the one with the smallest loss.

**The maths**

For every entry $i$ we have a label $y_i$, the true value, and a prediction $y_i^{\text{pred}}$, the output of the model. The loss is the sum of the squared differences between them:

$$L = \sum_{i} \left( y_i - y_i^{\text{pred}} \right)^2$$

The smaller the loss, the closer the predictions are to the labels.

Here $i$ runs over the 11 measurements. The prediction is a straight line with slope $a$ and intercept $b$, evaluated at each dose $x_i$:

$$y_i^{\text{pred}} = a\,x_i + b$$

Fitting the line means finding the slope and intercept that make the loss as small as possible: $(a^*, b^*) = \arg\min_{a, b} L$.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Input: doses of the stimulus
x = np.linspace(0, 10, 11)

# Label: measured gene expression (a hidden straight line plus some noise)
np.random.seed(0)
y = 0.5 * x + 1.2 + np.random.normal(0, 0.4, size = x.shape)

print("x:", x)
print("y:", np.round(y, 2))
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# Plot the measurements
plt.figure(figsize = (8, 5))
plt.scatter(x, y, color = 'blue', label = 'Measurements (label y)')
plt.xlabel('x: stimulus (input)')
plt.ylabel('y: gene expression')
plt.title('Experimental data')
plt.legend()
plt.show()
```

```python
# Model: a straight line. Input x -> output y_pred
def model(x, a, b):
    return a * x + b
```

```python
# Loss: sum of squared differences between the label y and the prediction y_pred
def loss(y, y_pred):
    return np.sum((y - y_pred)**2)
```

```python
# Table: input, label, output and squared error for each measurement
def print_table(x, y, y_pred):
    print(" input x | label y | output y_pred | (y - y_pred)^2")
    print("-" * 50)
    for xi, yi, yhi in zip(x, y, y_pred):
        print(f"  {xi:5.1f}  |  {yi:5.2f}  |    {yhi:5.2f}     |     {(yi - yhi)**2:5.2f}")
    print("-" * 50)
    print(f"Loss (sum of the last column): {loss(y, y_pred):.3f}")

# Plot: the line with its errors (left) and the squared error of each point (right)
def plot_loss(x, y, y_pred, title):
    fig, axes = plt.subplots(1, 2, figsize = (14, 5))

    axes[0].scatter(x, y, color = 'blue', label = 'Label y')
    axes[0].plot(x, y_pred, color = 'red', marker = 'o', label = 'Output y_pred')
    for xi, yi, yhi in zip(x, y, y_pred):
        axes[0].vlines(xi, yhi, yi, color = 'gray', linestyle = 'dashed')
    axes[0].set_xlabel('x: stimulus (input)')
    axes[0].set_ylabel('y: gene expression')
    axes[0].set_title(title)
    axes[0].legend()

    axes[1].bar(x, (y - y_pred)**2, color = 'gray')
    axes[1].set_xlabel('x: stimulus (input)')
    axes[1].set_ylabel('(y - y_pred)^2')
    axes[1].set_title(f'Error of each measurement. Loss = sum of the bars = {loss(y, y_pred):.3f}')

    plt.tight_layout()
    plt.show()
```

```python
# First guess: slope 1, intercept 0
a, b = 1.0, 0.0
y_pred = model(x, a, b)
print_table(x, y, y_pred)
plot_loss(x, y, y_pred, f'Guess: y_pred = {a} x + {b}')
```

```python
# Second guess: slope 0.5, intercept 1
a, b = 0.5, 1.0
y_pred = model(x, a, b)
print_table(x, y, y_pred)
plot_loss(x, y, y_pred, f'Guess: y_pred = {a} x + {b}')
```

```python
# Best line: the slope and intercept that minimize the loss (least squares)
# slope = sum((x - x_mean) * (y - y_mean)) / sum((x - x_mean)^2)
# intercept = y_mean - slope * x_mean
x_mean = np.mean(x)
y_mean = np.mean(y)
a_best = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean)**2)
b_best = y_mean - a_best * x_mean

y_pred = model(x, a_best, b_best)
print_table(x, y, y_pred)
plot_loss(x, y, y_pred, f'Best fit: y_pred = {a_best:.2f} x + {b_best:.2f}')
```

```python
# Check with numpy implementation
a_np, b_np = np.polyfit(x, y, deg = 1)
print(f"By hand: y_pred = {a_best:.3f} x + {b_best:.3f}")
print(f"numpy:   y_pred = {a_np:.3f} x + {b_np:.3f}")

# Summary: the best fit has the smallest loss
print(f"\nLoss of guess 1:  {loss(y, model(x, 1.0, 0.0)):.3f}")
print(f"Loss of guess 2:  {loss(y, model(x, 0.5, 1.0)):.3f}")
print(f"Loss of best fit: {loss(y, model(x, a_best, b_best)):.3f}")
```
