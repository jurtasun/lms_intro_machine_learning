**Unsupervised learning: clustering**

We have cells described by the expression of two genes, but no labels: we do not know which cell belongs to which population.
K-means splits the cells into $k$ groups by itself, and we check the result with scikit-learn.

**The maths**

Each group $j$ has a centre $\mu_j$, the average of the cells in that group. A good grouping has every cell close to its own centre.

We add up the squared distance from every cell $x_i$ to the centre of its group. This is the within-cluster sum of squares (WCSS), and k-means makes it as **small** as possible:

$$\text{WCSS} = \sum_{j=1}^{k} \; \sum_{x_i \text{ in group } j} \lVert x_i - \mu_j \rVert^2$$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Generate example input
# Rows represent cells, columns represent two genes
# We simulate three groups of cells, but we do not give the labels to the algorithm
np.random.seed(0)
group_1 = np.random.normal([2, 2], 0.5, size = (20, 2))
group_2 = np.random.normal([6, 2], 0.5, size = (20, 2))
group_3 = np.random.normal([4, 6], 0.5, size = (20, 2))
data = np.vstack([group_1, group_2, group_3])
print("Data (first 5 cells):\n", data[:5])
print("Shape:", data.shape)
print("Format: ", type(data))
```

```python
# Plot the data: no colours, we do not know the groups
plt.figure(figsize = (6, 6))
plt.scatter(data[:, 0], data[:, 1], color = 'gray')
plt.xlabel('gene 1')
plt.ylabel('gene 2')
plt.title('Cells without labels')
plt.show()
```

```python
# Number of groups we look for
k = 3

# Start with k random cells as the centroids (centre of each group)
start = np.random.choice(len(data), size = k, replace = False)
centroids = data[start]
print("Initial centroids:\n", centroids)
```

```python
# K-means implementation
# Repeat two steps:
# 1. Assign each cell to its closest centroid
# 2. Move each centroid to the mean of its cells
print("K-means running")

for step in range(10):

    # Euclidean distance from every cell to every centroid
    distances = np.zeros((len(data), k))
    for j in range(k):
        distances[:, j] = np.sqrt(np.sum((data - centroids[j])**2, axis = 1))

    # Step 1: closest centroid for each cell
    clusters = np.argmin(distances, axis = 1)

    # Step 2: new centroid = mean of the cells in that cluster
    for j in range(k):
        centroids[j] = np.mean(data[clusters == j], axis = 0)

    # Within-cluster sum of squares: how compact the groups are
    # This is what k-means makes smaller at every step
    wcss = np.sum(np.min(distances, axis = 1)**2)
    print(f"Step {step}: WCSS = {wcss:.2f}")

print("\nFinal centroids:\n", centroids)
```

```python
# Plot the groups found by k-means
plt.figure(figsize = (6, 6))
plt.scatter(data[:, 0], data[:, 1], c = clusters, cmap = 'viridis')
plt.scatter(centroids[:, 0], centroids[:, 1], color = 'red', marker = 'x', s = 150, label = 'Centroids')
plt.xlabel('gene 1')
plt.ylabel('gene 2')
plt.title('Groups found by k-means')
plt.legend()
plt.show()
```

```python
# Check with scikit-learn implementation
from sklearn.cluster import KMeans

model = KMeans(n_clusters = 3, n_init = 10, random_state = 0).fit(data)
print("scikit-learn centroids:\n", model.cluster_centers_)
print(f"scikit-learn WCSS: {model.inertia_:.2f}")
```
