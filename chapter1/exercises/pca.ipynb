**Unsupervised learning: dimensionality reduction**

We have cells described by five genes, too many to plot at once, and no labels.
PCA finds two new variables (principal components) that keep most of the information, so we can see the cells in a 2D plot.

**The maths**

The covariance matrix $\Sigma$ tells us how the genes vary together. The first principal component is the direction $w$ along which the cells spread the **most**, that is, with the largest variance:

$$\max_{w} \; w^\top \Sigma \, w \qquad \text{with } \lVert w \rVert = 1$$

The solution are the eigenvectors of $\Sigma$. Each eigenvalue $\lambda$ is the variance along its direction, so we sort them from largest to smallest:

$$\Sigma \, w = \lambda \, w$$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Generate example input
# Rows represent cells, columns represent 5 genes
# We simulate three cell populations, but we do not give the labels to PCA
np.random.seed(0)
population_1 = np.random.normal([2, 8, 5, 1, 4], 0.7, size = (20, 5))
population_2 = np.random.normal([6, 3, 5, 4, 4], 0.7, size = (20, 5))
population_3 = np.random.normal([8, 7, 2, 6, 4], 0.7, size = (20, 5))
data = np.vstack([population_1, population_2, population_3])
print("Data (first 5 cells):\n", np.round(data[:5], 2))
print("Shape:", data.shape)
print("Format: ", type(data))
```

```python
# With 5 genes we cannot plot all of them at once
# Plotting only two genes shows part of the picture
plt.figure(figsize = (8, 5))
plt.scatter(data[:, 0], data[:, 1], color = 'gray')
plt.xlabel('gene 1')
plt.ylabel('gene 2')
plt.title('Only two of the five genes')
plt.show()
```

```python
# Step 1: centre the data (subtract the mean of each gene)
data_centred = data - np.mean(data, axis = 0)
print("Mean of each gene after centring:", np.round(np.mean(data_centred, axis = 0), 2))
```

```python
# Step 2: covariance matrix, how genes vary together
# Shape (5, 5): one row and one column per gene
cov = np.cov(data_centred, rowvar = False)
print("Covariance matrix:\n", np.round(cov, 2))
print("Shape:", cov.shape)
```

```python
# Step 3: eigenvectors and eigenvalues of the covariance matrix
# Eigenvectors: directions of the principal components
# Eigenvalues: variance along each direction
eigenvalues, eigenvectors = np.linalg.eigh(cov)

# Sort from largest to smallest variance
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]
print("Eigenvalues:", np.round(eigenvalues, 2))
```

```python
# Explained variance: fraction of the total variance in each component
explained = eigenvalues / np.sum(eigenvalues)
print("Explained variance:", np.round(explained, 2))
print(f"PC1 + PC2 explain {np.sum(explained[:2]):.0%} of the variance")
```

```python
# Step 4: project the cells onto the first two components
pcs = data_centred @ eigenvectors[:, :2]
print("PCs (first 5 cells):\n", np.round(pcs[:5], 2))
print("Shape:", pcs.shape)
```

```python
# Plot the cells in the PC1 - PC2 plane
# We never used the labels, but the three populations appear as groups
plt.figure(figsize = (8, 5))
plt.scatter(pcs[:, 0], pcs[:, 1], color = 'gray')
plt.xlabel(f'PC1 ({explained[0]:.0%} of variance)')
plt.ylabel(f'PC2 ({explained[1]:.0%} of variance)')
plt.title('PCA: cells without labels')
plt.show()
```

```python
# Check with scikit-learn implementation
# The sign of each component can be flipped, it is the same direction
from sklearn.decomposition import PCA

model = PCA(n_components = 2).fit(data)
print("scikit-learn explained variance:", np.round(model.explained_variance_ratio_, 2))
print("Our explained variance:         ", np.round(explained[:2], 2))
```
