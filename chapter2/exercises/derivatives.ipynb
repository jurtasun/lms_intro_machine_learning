**Derivatives**

To make a loss smaller we need to know in which direction it goes down. The derivative gives the slope of a function at each point: how much $f$ changes when $x$ changes a little.
We write the most common functions and their derivatives, plot them, and check the derivatives numerically with numpy.

**The maths**

The derivative of $f$ at a point $x_0$ is the slope of $f$ at that point: the change in $f$ when we take a very small step $\Delta x$ away from $x_0$, divided by the size of the step:

$$f'(x_0) = \lim_{\Delta x \to 0} \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x}$$

Doing this at every point $x$ gives a new function, the derivative $f'(x)$. Where $f'(x) > 0$ the function goes up, where $f'(x) < 0$ it goes down, and at a minimum or maximum the slope is zero, $f'(x) = 0$. This is what we use to minimize a loss.

Examples of the most common functions and their derivatives:

| function $f(x)$ | derivative $f'(x)$ |
|---|---|
| $3$ (constant) | $0$ |
| $2x + 1$ | $2$ |
| $x^2$ | $2x$ |
| $x^3$ | $3x^2$ |
| $a^x$ | $a^x \ln(a)$ |
| $e^x$ | $e^x$ |

The exponential $e^x$ is the special case $a = e$: since $\ln(e) = 1$, it stays the same when we take its derivative.
| $\sin(x)$ | $\cos(x)$ |
| $\cos(x)$ | $-\sin(x)$ |

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Grid of x values
x = np.linspace(-3, 3, 200)
print("x (first 5 values):", x[:5])
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# Plot a function (left) and its derivative (right)
def plot_function(f, df, f_name, df_name):
    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    axes[0].plot(x, f(x))
    axes[0].set_title(f_name)
    axes[0].set_xlabel("x")
    axes[0].set_ylabel("f(x)")
    axes[0].grid(True)

    axes[1].plot(x, df(x), color = 'red')
    axes[1].axhline(0, color = 'gray', linestyle = 'dashed') # slope = 0
    axes[1].set_title(df_name)
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("f'(x)")
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()
```

```python
# Constant: it never changes, so the slope is zero everywhere
def constant(x):
    return 3 * np.ones_like(x)

def constant_derivative(x):
    return np.zeros_like(x)

plot_function(constant, constant_derivative, "f(x) = 3", "f'(x) = 0")
```

```python
# Straight line: the slope is the same everywhere
def line(x):
    return 2 * x + 1

def line_derivative(x):
    return 2 * np.ones_like(x)

plot_function(line, line_derivative, "f(x) = 2x + 1", "f'(x) = 2")
```

```python
# Parabola: slope negative on the left, positive on the right, zero at the minimum
def parabola(x):
    return x**2

def parabola_derivative(x):
    return 2 * x

plot_function(parabola, parabola_derivative, "f(x) = x^2", "f'(x) = 2x")
```

```python
# Cubic: always goes up, the slope is zero only at x = 0
def cubic(x):
    return x**3

def cubic_derivative(x):
    return 3 * x**2

plot_function(cubic, cubic_derivative, "f(x) = x^3", "f'(x) = 3x^2")
```

```python
# Power of a number a: a^x has derivative a^x * ln(a). Try other values of a
a = 2

def power(x):
    return a**x

def power_derivative(x):
    return a**x * np.log(a)

plot_function(power, power_derivative, f"f(x) = {a}^x", f"f'(x) = {a}^x ln({a})")
```

```python
# Exponential: the case a = e. Since ln(e) = 1, its slope is the function itself
plot_function(np.exp, np.exp, "f(x) = exp(x)", "f'(x) = exp(x)")
```

```python
# Sine: its slope is the cosine
plot_function(np.sin, np.cos, "f(x) = sin(x)", "f'(x) = cos(x)")
```

```python
# Cosine: its slope is minus the sine
def minus_sin(x):
    return -np.sin(x)

plot_function(np.cos, minus_sin, "f(x) = cos(x)", "f'(x) = -sin(x)")
```

```python
# Check with numpy implementation
# np.gradient approximates the slope from neighbouring points: (f(x + dx) - f(x - dx)) / (2 dx)
functions = [
    ("3", constant, constant_derivative),
    ("2x + 1", line, line_derivative),
    ("x^2", parabola, parabola_derivative),
    ("x^3", cubic, cubic_derivative),
    (f"{a}^x", power, power_derivative),
    ("exp(x)", np.exp, np.exp),
    ("sin(x)", np.sin, np.cos),
    ("cos(x)", np.cos, minus_sin),
]

for name, f, df in functions:
    numerical = np.gradient(f(x), x)
    difference = np.max(np.abs(numerical[1:-1] - df(x)[1:-1]))  # skip the two end points
    print(f"{name:8s} max difference: {difference:.5f}")
```
