**Activation functions**

A neuron adds up its inputs and passes the result through an activation function, which decides how strongly it fires.
We implement ReLU, sigmoid and softmax by hand, plot them, and check them with TensorFlow and PyTorch.

**The maths**

ReLU (rectified linear unit), lets positive values through and sets negative values to zero:

$$\text{ReLU}(x) = \max(0, x)$$

Sigmoid, squeezes any number into a value between 0 and 1, like a probability:

$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

Softmax, turns a list of numbers $x_1, \dots, x_n$ into probabilities that add up to 1, giving more weight to the larger values:

$$\text{softmax}(x)_i = \frac{e^{x_i}}{\sum_{j=1}^{n} e^{x_j}}$$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import torch

# Check installation
print("All libraries imported successfully")
print(np.__version__)
print(tf.__version__)
print(torch.__version__)
```

```python
# Generate example input data
x = np.array([-1.0, 0.0, 1.0, 2.0, 3.0])
print("\nExplore input:")
print("x: ", x)
print("x shape:", x.shape)
print("x format: ", type(x))
```

```python
# Manual implementation

# ReLU
def relu_manual(x):
    return np.maximum(0, x)

# Sigmoid
def sigmoid_manual(x):
    return 1 / (1 + np.exp(-x))

# Softmax
def softmax_manual(x):
    exps = np.exp(x - np.max(x)) # for numerical stability
    return exps / np.sum(exps)

# Check manual implementation
sigmoid_m = sigmoid_manual(x)
relu_m = relu_manual(x)
softmax_m = softmax_manual(x)
print("\nManual implementation:")
print("Sigmoid:", sigmoid_m)
print("ReLU:", relu_m)
print("Softmax:", softmax_m)
```

```python
# Plot activation functions

# Prepare grid and figure panel
x_plot = np.linspace(-5, 5, 200)
fig, axes = plt.subplots(1, 3, figsize = (15, 4))

# Plot ReLU(x)
axes[0].plot(x_plot, relu_manual(x_plot))
axes[0].set_title("ReLU")
axes[0].set_xlabel("x")
axes[0].set_ylabel("relu(x)")
axes[0].grid(True)

# Plot Sigmoid(x)
axes[1].plot(x_plot, sigmoid_manual(x_plot))
axes[1].set_title("Sigmoid")
axes[1].set_xlabel("x")
axes[1].set_ylabel("sigmoid(x)")
axes[1].grid(True)

# Plot Softmax(x)
axes[2].bar(range(len(x)), softmax_m)
axes[2].set_xticks(range(len(x)))
axes[2].set_xticklabels(x)
axes[2].set_title("Softmax")
axes[2].set_xlabel("x")
axes[2].set_ylabel("softmax(x)")
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
# TensorFlow implementation

# Convert input to required format
x_tf = tf.constant(x, dtype = tf.float32)

# Check tensorflow implementation
sigmoid_tf = tf.nn.sigmoid(x_tf).numpy()
relu_tf = tf.nn.relu(x_tf).numpy()
softmax_tf = tf.nn.softmax(x_tf).numpy()
print("\nTensorFlow implementation:")
print("Sigmoid:", sigmoid_tf)
print("ReLU:", relu_tf)
print("Softmax:", softmax_tf)
```

```python
# PyTorch implementation

# Convert input to required format
x_torch = torch.tensor(x, dtype = torch.float32)

# Check pytorch implementation
sigmoid_torch = torch.sigmoid(x_torch).numpy()
relu_torch = torch.relu(x_torch).numpy()
softmax_torch = torch.softmax(x_torch, dim=0).numpy()
print("\nPyTorch implementation:")
print("Sigmoid:", sigmoid_torch)
print("ReLU:", relu_torch)
print("Softmax:", softmax_torch)
```

```python
# Precision check (rtol = 1e-5)

print("\nPrecision check (Manual vs TensorFlow):")
print("Sigmoid close: ", np.allclose(sigmoid_m, sigmoid_tf))
print("ReLU close: ", np.allclose(relu_m, relu_tf))
print("Softmax close: ", np.allclose(softmax_m, softmax_tf))

print("\nCross-check (Manual vs PyTorch):")
print("Sigmoid close: ", np.allclose(sigmoid_m, sigmoid_torch))
print("ReLU close: ", np.allclose(relu_m, relu_torch))
print("Softmax close: ", np.allclose(softmax_m, softmax_torch))
```
