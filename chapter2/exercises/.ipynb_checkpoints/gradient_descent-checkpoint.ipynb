**Gradient descent**

We want the slope of the line that best fits gene expression vs stimulus. Instead of using a formula, we start from a guess and walk downhill on the loss, step by step.
The derivative of the loss tells us in which direction to move. We check the result with the exact least squares solution.

**The maths**

The model is a line through the origin with slope $a$, and the loss is the mean squared error between observations $y_i$ and predictions $a\,x_i$:

$$L(a) = \frac{1}{N} \sum_{i=1}^{N} \left( y_i - a\,x_i \right)^2$$

Its derivative with respect to $a$ tells us if the loss goes up or down when we increase $a$:

$$\frac{dL}{da} = -\frac{2}{N} \sum_{i=1}^{N} x_i \left( y_i - a\,x_i \right)$$

Gradient descent moves $a$ a small step against the slope, so the loss goes down. The learning rate $\eta$ sets the size of the step:

$$a \leftarrow a - \eta \, \frac{dL}{da}$$

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Example data: gene expression y at different doses of a stimulus x
np.random.seed(0)
x = np.linspace(0, 10, 11)
y = 0.5 * x + np.random.normal(0, 0.4, size = x.shape)
print("x:", x)
print("y:", np.round(y, 2))
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# Loss and its derivative
def loss(a):
    return np.mean((y - a * x)**2)

def loss_derivative(a):
    return -2 * np.mean(x * (y - a * x))

print(f"Loss at a = 0:   {loss(0):.3f}, derivative: {loss_derivative(0):.3f}")
print(f"Loss at a = 1:   {loss(1):.3f}, derivative: {loss_derivative(1):.3f}")
```

```python
# Plot the loss as a function of the slope
# Negative derivative: the loss goes down if we increase a. Positive: it goes down if we decrease a
a_grid = np.linspace(-0.5, 1.5, 200)
plt.figure(figsize = (8, 5))
plt.plot(a_grid, [loss(a) for a in a_grid], color = 'purple')
plt.xlabel('slope a')
plt.ylabel('loss L(a)')
plt.title('Loss vs slope')
plt.grid(True)
plt.show()
```

```python
# Gradient descent implementation
# Try learning_rate = 0.001 (slow) or 0.03 (the steps are too big and it jumps away)
learning_rate = 0.01
a = 0.0          # initial guess
path = [a]

print("Gradient descent running")
for step in range(15):
    a = a - learning_rate * loss_derivative(a)
    path.append(a)
    print(f"Step {step}: a = {a:.4f}, loss = {loss(a):.4f}")
```

```python
# Plot the steps on top of the loss curve
path = np.array(path)
plt.figure(figsize = (8, 5))
plt.plot(a_grid, [loss(a) for a in a_grid], color = 'purple', label = 'Loss')
plt.plot(path, [loss(a) for a in path], color = 'red', marker = 'o', label = 'Gradient descent steps')
plt.xlabel('slope a')
plt.ylabel('loss L(a)')
plt.title('Walking downhill on the loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Plot the final line and the observations
plt.figure(figsize = (8, 5))
plt.scatter(x, y, color = 'blue', label = 'Observations')
plt.plot(x, a * x, color = 'red', label = f'Fitted line: y = {a:.2f} x')
plt.xlabel('x: stimulus')
plt.ylabel('y(x): gene expression')
plt.title('Line found by gradient descent')
plt.legend()
plt.show()
```

```python
# Check with the exact least squares solution
# Setting dL/da = 0 gives a = sum(x * y) / sum(x^2)
a_exact = np.sum(x * y) / np.sum(x**2)
print(f"Gradient descent: a = {a:.4f}")
print(f"Exact solution:   a = {a_exact:.4f}")
print("Close: ", np.isclose(a, a_exact))
```
