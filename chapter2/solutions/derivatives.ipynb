**Derivatives**

To make a loss smaller we need to know in which direction it goes down. The derivative gives the slope of a function at each point: how much $f$ changes when $x$ changes a little.
We write the most common functions and their derivatives, plot them, and check the derivatives numerically with numpy.

**The maths**

The derivative of $f$ at a point $x_0$ is the slope of $f$ at that point: the change in $f$ when we take a very small step $\Delta x$ away from $x_0$, divided by the size of the step:

$$f'(x_0) = \lim_{\Delta x \to 0} \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x}$$

Doing this at every point $x$ gives a new function, the derivative $f'(x)$. Where $f'(x) > 0$ the function goes up, where $f'(x) < 0$ it goes down, and at a minimum or maximum the slope is zero, $f'(x) = 0$. This is what we use to minimize a loss.

Examples of the most common functions and their derivatives:

| function $f(x)$ | derivative $f'(x)$ |
|---|---|
| $3$ (constant) | $0$ |
| $2x + 1$ | $2$ |
| $x^2$ | $2x$ |
| $x^3$ | $3x^2$ |
| $a^x$ | $a^x \ln(a)$ |
| $e^x$ | $e^x$ |

The exponential $e^x$ is the special case $a = e$: since $\ln(e) = 1$, it stays the same when we take its derivative.
| $\sin(x)$ | $\cos(x)$ |
| $\cos(x)$ | $-\sin(x)$ |

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Grid of x values
x = np.linspace(-3, 3, 200)

print("x (first 5 values):", x[:5])
print("x (last 5 values):", x[-5:])
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# Small step for the manual derivative: the Δx of the definition
dx = 1e-10

# Plot a function (left) and its derivative computed with the definition (right)
def plot_function(f, df_manual, f_name):

    fig, axes = plt.subplots(1, 2, figsize = (12, 4))

    axes[0].plot(x, f(x))
    axes[0].set_title(f_name)
    axes[0].set_xlabel("x")
    axes[0].set_ylabel("f(x)")
    axes[0].grid(True)

    axes[1].plot(x, df_manual(x), color = 'red')
    axes[1].set_title(f"derivative of {f_name}")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("f'(x)")
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()
```

```python
# Constant: it never changes, so the slope is zero everywhere
def constant(x):
    return 3 * np.ones_like(x)

def constant_derivative(x):
    return (constant(x + dx) - constant(x)) / dx   # definition: (f(x + Δx) - f(x)) / Δx

plot_function(constant, constant_derivative, "f(x) = 3")
```

```python
# Straight line: the slope is the same everywhere
def line(x):
    return 2 * x + 1

def line_derivative(x):
    return (line(x + dx) - line(x)) / dx

plot_function(line, line_derivative, "f(x) = 2x + 1")
```

```python
# Parabola: slope negative on the left, positive on the right, zero at the minimum
def parabola(x):
    return x**2

def parabola_derivative(x):
    return (parabola(x + dx) - parabola(x)) / dx

plot_function(parabola, parabola_derivative, "f(x) = x^2")
```

```python
# Cubic: always goes up, the slope is zero only at x = 0
def cubic(x):
    return x**3

def cubic_derivative(x):
    return (cubic(x + dx) - cubic(x)) / dx

plot_function(cubic, cubic_derivative, "f(x) = x^3")
```

```python
# Power of a number a. Try other values of a
a = 2

def power(x):
    return a**x

def power_derivative(x):
    return (power(x + dx) - power(x)) / dx

plot_function(power, power_derivative, f"f(x) = {a}^x")
```

```python
# Exponential: the case a = e. Its slope looks exactly like the function itself
def exp_derivative(x):
    return (np.exp(x + dx) - np.exp(x)) / dx

plot_function(np.exp, exp_derivative, "f(x) = exp(x)")
```

```python
# Sine: its slope looks like a cosine
def sin_derivative(x):
    return (np.sin(x + dx) - np.sin(x)) / dx

plot_function(np.sin, sin_derivative, "f(x) = sin(x)")
```

```python
# Cosine: its slope looks like minus a sine
def cos_derivative(x):
    return (np.cos(x + dx) - np.cos(x)) / dx

plot_function(np.cos, cos_derivative, "f(x) = cos(x)")
```

```python
# Final check: is our manual derivative correct?

# We have computed every derivative ourselves, with the definition:
#     slope ≈ (f(x + dx) - f(x)) / dx      → one small step to the right

# numpy can also estimate the slope, with its own function np.gradient()
# It does it slightly differently: it looks one point to the left and one to the right.

# If both methods give (almost) the same numbers, our manual derivative is right.

# List of (name, function, our manual derivative) to check
functions = [
    ("3", constant, constant_derivative),
    ("2x + 1", line, line_derivative),
    ("x^2", parabola, parabola_derivative),
    ("x^3", cubic, cubic_derivative),
    (f"{a}^x", power, power_derivative),
    ("exp(x)", np.exp, exp_derivative),
    ("sin(x)", np.sin, sin_derivative),
    ("cos(x)", np.cos, cos_derivative),
]

print("function | biggest difference | do they agree?")
print("-" * 46)

for name, f, df_manual in functions:

    # our slope at every point of x
    ours = df_manual(x)

    # numpy's slope at every point of x
    numpy_slope = np.gradient(f(x), x) 

    # Distance between the two, point by point; keep only the largest one
    # (we skip the first and last point, where numpy can only look to one side)
    difference = np.max(np.abs(ours[1:-1] - numpy_slope[1:-1]))

    agree = "yes" if difference < 0.01 else "no"
    print(f"{name:8s} |      {difference:.5f}       |  {agree}")
```
