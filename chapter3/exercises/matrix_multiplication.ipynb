**Weights and biases: matrix multiplication**

A neuron multiplies each input by a weight, adds them up, and adds a bias. A layer has many neurons, each with its own weights.
Writing all the neurons at once is a matrix times a vector. We do it by hand with loops and check it with numpy.

**The maths**

One neuron with inputs $x_1, x_2, x_3$ (e.g. the expression of three genes), weights $w_1, w_2, w_3$ and bias $b$:

$$z = w_1 x_1 + w_2 x_2 + w_3 x_3 + b$$

A layer of neurons puts the weights of each neuron in one row of a matrix $W$, and the biases in a vector $b$. For 2 neurons and 3 inputs, $W$ has 2 rows and 3 columns:

$$z = W x + b \qquad \begin{pmatrix} z_1 \\ z_2 \end{pmatrix} = \begin{pmatrix} w_{11} & w_{12} & w_{13} \\ w_{21} & w_{22} & w_{23} \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix} + \begin{pmatrix} b_1 \\ b_2 \end{pmatrix}$$

```python
# Import libraries
import numpy as np

# Check installation
print("All libraries imported successfully")
print(np.__version__)
```

```python
# Input: expression of 3 genes in one cell
x = np.array([1.0, 2.0, 3.0])
print("x:", x)
print("Shape:", x.shape)
print("Format: ", type(x))
```

```python
# One neuron: one weight per input, plus a bias
w = np.array([0.5, -1.0, 2.0])
b = 1.0

# By hand: multiply each input by its weight, add them up, add the bias
z = w[0] * x[0] + w[1] * x[1] + w[2] * x[2] + b
print("Neuron output (by hand):", z)

# Same with numpy: dot product
print("Neuron output (np.dot): ", np.dot(w, x) + b)
```

```python
# A layer with 2 neurons: one row of weights per neuron
W = np.array([
    [0.5, -1.0, 2.0],   # neuron 1
    [1.0,  0.0, -0.5]]) # neuron 2
b = np.array([1.0, 2.0])
print("W:\n", W)
print("Shape:", W.shape)
print("b:", b)
print("Shape:", b.shape)
```

```python
# Matrix times vector by hand: each neuron does its own weighted sum
z_loop = np.zeros(2)
for i in range(2):          # neurons (rows of W)
    for j in range(3):      # inputs (columns of W)
        z_loop[i] += W[i, j] * x[j]
    z_loop[i] += b[i]
print("Layer output (loops):", z_loop)
print("Shape:", z_loop.shape)
```

```python
# Check with numpy implementation: the @ symbol is matrix multiplication
z_numpy = W @ x + b
print("Layer output (numpy):", z_numpy)
print("Close: ", np.allclose(z_loop, z_numpy))
```

```python
# Many cells at once: one row per cell
X = np.array([
    [1.0, 2.0, 3.0],
    [0.5, 0.1, 2.0],
    [3.0, 1.0, 0.0],
    [2.0, 2.0, 2.0]])
print("X shape (cells, genes):", X.shape)

# Each row of the output is the layer applied to one cell
Z = X @ W.T + b
print("Z:\n", Z)
print("Z shape (cells, neurons):", Z.shape)
```
